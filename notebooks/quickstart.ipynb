# Spectral-KRR quickstart
Spectral filtering of the FC matrix, then a ridge comparison on a small training set.

```python
import sys; sys.path.insert(0, '../src')
import numpy as np
from data import load_abide, target_config, site_indices, train_test_split
from spectral import build_cache
from utils import fit_ridge, metrics_reg
from sklearn.preprocessing import StandardScaler

data = load_abide('../data/abide_cc200_asd.npz')
cache = build_cache(data['X'], [0.3])
```

```python
t = target_config(data)['fiq']
hold = site_indices(t['valid'], data['site'], 'NYU')
y = t['y'][hold]
tr, te = train_test_split(len(y), 100, seed=0)

for name, F in [('vanilla KRR', data['X']), ('Spectral-KRR', cache[0.3])]:
    sc = StandardScaler().fit(F[hold[tr]])
    mu = fit_ridge(sc.transform(F[hold[tr]]), y[tr], sc.transform(F[hold[te]]))
    print(name, metrics_reg(y[te], mu))
```
